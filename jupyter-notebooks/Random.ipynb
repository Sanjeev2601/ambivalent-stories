```python
import pandas as pd
from datetime import datetime
import os

# Paths
BASE_DIR = "../data/OP_Replies/dataset"
PARENT_FILE = os.path.join(BASE_DIR, "Ambivalent_Data_OP_Replies.csv")
SAMPLE_SIZE = 287

# Load parent CSV (header automatically preserved)
df = pd.read_csv(PARENT_FILE)

# Safety check
if len(df) < SAMPLE_SIZE:
    raise ValueError(f"Not enough rows to sample {SAMPLE_SIZE}. Rows left: {len(df)}")

# Randomly sample 50 rows
sampled_df = df.sample(n=SAMPLE_SIZE, random_state=None)

# Remove sampled rows from parent
remaining_df = df.drop(sampled_df.index)

# Create timestamped output filename
timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
output_file = os.path.join(
    BASE_DIR,
    f"Ambivalent_Random_Round_6.csv"
)

# Save files WITH headers
sampled_df.to_csv(output_file, index=False)
remaining_df.to_csv(PARENT_FILE, index=False)

print("✅ Sampling complete")
print(f"📄 Created: {output_file}")
print(f"📉 Remaining rows in parent file: {len(remaining_df)}")
```
