```python
import pandas as pd
from sklearn.metrics import cohen_kappa_score

# =========================
# UPDATE PATH
# =========================
FILE = "Round2_50.csv"

# =========================
# LOAD FILE
# =========================
df = pd.read_csv(FILE)

# =========================
# BASE LABEL NAMES
# =========================
LABELS = [
    "Question",
    "Answer",
    "Announcement",
    "Agreement",
    "Appreciation",
    "Disagreement",
    "Negative Reaction",
    "Elaboration",
    "Humor"
]

# =========================
# NORMALIZE yes/no → 1/0
# =========================
def normalize(series):
    return series.astype(str).str.strip().str.lower().map({
        "yes": 1,
        "no": 0
    })

# =========================
# COMPUTE KAPPA
# =========================
results = []

for label in LABELS:
    col_1 = f"{label}_Sanjeev"
    col_2 = f"{label}_Shirlene"

    y1 = normalize(df[col_1])
    y2 = normalize(df[col_2])

    # Keep only rows where both annotators labeled
    valid = y1.notna() & y2.notna()
    y1 = y1[valid]
    y2 = y2[valid]

    kappa = cohen_kappa_score(y1, y2)

    results.append({
        "Label": label,
        "Cohen_Kappa": round(kappa, 4),
        "Support": len(y1)
    })

kappa_df = pd.DataFrame(results)
kappa_df
```
